```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/List_of_languages_by_number_of_native_speakers_in_India"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
}

response = requests.get(url, headers=headers, timeout=20)

print(response.status_code)
print(response.url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.title.text)
```

```python
title = soup.title.text.replace(" - Wikipedia", "")
print(title)
```

```python
tables = soup.find_all("table")

print(len(tables))
```

```python
for i, table in enumerate(tables):
    print("TABLE", i)
    print(table.get_text(" ", strip=True)[:300])
    print("-" * 50)
```

```python
language_table = tables[1]
```

```python
rows = language_table.find_all("tr")

for row in rows[:5]:
    print(row.get_text(" | ", strip=True))
```

```python
data = []

for row in rows[1:]:
    cells = row.find_all(["th", "td"])
    cells = [cell.get_text(" ", strip=True) for cell in cells]
    
    if cells:
        data.append(cells)
```

```python
for row in data[:5]:
    print(row)
```

```python
for row in data[:5]:
    print(len(row))
```

```python
for row in data[:3]:
    print(row)
```

```python
language_data = []

for row in data[1:]:
    language_data.append({
        "rank": row[0],
        "language": row[1],
        "speakers_2011": row[6],
        "percentage_2011": row[7]
    })

language_df = pd.DataFrame(language_data)
```

```python
language_df.head()
```

```python
language_df["language"] = language_df["language"].str.replace(
    r"\[.*?\]",
    "",
    regex=True
).str.strip()
```

```python
language_df.head()
```

```python
language_df = pd.DataFrame(data[1:], columns=[
    "rank",
    "language",
    "speakers_1991",
    "percentage_1991",
    "speakers_2001",
    "percentage_2001",
    "speakers_2011",
    "percentage_2011",
    "worldwide_speakers"
])
```

```python
language_df.head()
```

```python
language_df["language"] = (
    language_df["language"]
    .str.replace(r"\[.*?\]", "", regex=True)
    .str.strip()
)
```

```python
language_df.to_csv("india_languages.csv", index=False)
```

```python
for col in ["percentage_1991", "percentage_2001", "percentage_2011"]:
    language_df[col] = (
        language_df[col]
        .astype(str)
        .str.replace("%", "", regex=False)
        .str.strip()
    )

    language_df[col] = pd.to_numeric(
        language_df[col],
        errors="coerce"
    )
```

```python
links = []

for row in rows[1:]:
    link = row.find("a")

    if link and link.get("href"):
        href = link.get("href")

        if href.startswith("/"):
            links.append("https://en.wikipedia.org" + href)
        else:
            links.append(href)
    else:
        links.append(None)
```
